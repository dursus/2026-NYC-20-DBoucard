###**Task 7:**

**Read the total profit of all months and show it using the Bar plot.**
* The generated **Bar plot** must include the following properties:
  * X label name = Month number.
  * Y label name = Total profit.
  * Title = Company profit per month.

The graph should look like the screenshot below:

Screenshot 2026-09-25 095737.png

```python
# Assign the x and y values using the loc reference
x_bar = df.loc[:,"month_number"]
y_bar = df.loc[:,"total_profit"]

# Create bar graph, try to match hex code and width of example
plt.bar(x_bar, y_bar, color="#1f77b4", width=0.8)

# Set labels and title
plt.xlabel("Month number")
plt.ylabel("Total profit")
plt.title("Company profit per month")

# Add a grid like the example shown where the grid is overlayed on top
plt.grid(linestyle = "dashed")
plt.gca().set_axisbelow(False)

# Display the plot
plt.show()
```

###**Task 8:**

**Read the total profit of all of the months, and show the line plot with the following style properties:**

* The generated **line plot** must include the following Style properties:
  * The line style should be **dotted** and the **line color** should be <font color="red">**red**</font>.
  * Show the legend at the lower-right location.
  * X label name = Month number
  * Y label name = Total profit.
  * Title = Company profit per month.
  * Add a circle marker; the line marker color = black.
  * The line width should be 3.

The graph should look like the screenshot below:

Screenshot 2026-09-25 100228.png

```python
# Assign x and y values
x_line = df.loc[:,"month_number"]
y_line = df.loc[:,"total_profit"]

# The instruction says dotted but the visual reference is dashed.
# If I wanted it dashed, I would replace the "dotted" with "dashed"
plt.plot(x_line,
         y_line,
         linestyle="dotted",
         color="red",
         linewidth=3,
         marker="o",
         markerfacecolor="black",
         label="Profit data of last year")

# Add labels, title, and legend to lower right
plt.xlabel("Month Number")
plt.ylabel("Total profit")
plt.title("Company profit per month")
plt.legend(loc="lower right")

# Display the plot
plt.show()
```

###**Task 9:**

**Print all of the product sales data and show it using a multi-line plot.**
* Display the number of units sold per month for each product using multi-line plots. (i.e., show a separate plotline for each product).
* The graph should look like the screenshot below:

Screenshot 2026-09-25 100351.png


```python
# Assign the x axis to month and all the y values for each category
month = df.loc[:,"month_number"]
face_cream = df.loc[:,"facecream"]
face_wash = df.loc[:,"facewash"]
toothpaste = df.loc[:,"toothpaste"]
bathing_soap = df.loc[:,"bathingsoap"]
shampoo = df.loc[:,"shampoo"]
moisturizer = df.loc[:,"moisturizer"]

# Add this line to give it more space so the lines are not squeezed together
plt.figure(figsize=(10, 10))

# Plot each line
plt.plot(month,face_cream,"-", color=("blue"), marker="o", label="Face cream Sales Data")
plt.plot(month,face_wash,"-", color=("yellow"), marker="o", label="Face Wash Sales Data")
plt.plot(month,toothpaste,"-", color=("green"), marker="o", label="Toothpaste Sales Data")
plt.plot(month,bathing_soap,"-", color=("red"), marker="o", label="Bathingsoap Sales Data")
plt.plot(month,shampoo,"-", color=("purple"), marker="o", label="Shampoo Sales Data")
plt.plot(month,moisturizer,"-", color=("brown"), marker="o", label="Moisturizer Sales Data")

# Add labels, title, and legend to upper left
plt.xlabel("Month number")
plt.ylabel("Sales units in number")
plt.title("Sales Data")
plt.legend(loc="upper left")

# Display the plot
plt.show()
```

###**Task 10:**

**Read “bathingsoap” sales data for each month and show it using a scatter plot.**
* Add a grid to the plot. The grid line style should be “–“.
* The scatter plot should look like the screenshot below:

Screenshot 2026-09-25 100458.png

```python
# Assign the values to x and y
x_scatter = np.array(df.loc[:,"month_number"])
y_scatter = np.array(df.loc[:,"bathingsoap"])

# Create the scatterplot and overlay a dashed grid on top to match the example
plt.scatter(x_scatter, y_scatter, label="bathing soap Sales data")
plt.grid(linestyle="dashed")
plt.gca().set_axisbelow(False)

# Add labels, title, and legend to top left
plt.xlabel("Month Number")
plt.ylabel("Number of units Sold")
plt.title("bathingsoap Sales Data")
plt.legend(loc="upper left")
plt.show()
```

###**Task 2:**

**Given Dataset:**

<font color="blue">height=[121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6]

weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]</font>

Show the average weight against the average height (use a line chart or any other chart):
* The generated **line plot** must include the following Style properties:
  * The line style should be dash-dot and the line color should be green.
  * Show the legend at the lower-right location.
  * X label name = Weight in kg.
  * Y label name = Height in cm.
  * Title = Average weight with respect to average height.
  * Add a circle marker: marker size = 10; circle marker color = green.

The graph should look like the screenshot below:

Screenshot 2026-09-25 101051.png

```python
# Assign the values to height and weight
height=[121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6]
weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]

# Plot the points and meet all the requirements as defined by this SBA
plt.plot(height,
         weight,
         linestyle="-.",
         color="green",
         marker="o",
         markersize=10,
         label="Weight to height")

# Add label, title, and legend to lower right
plt.xlabel("Weight in kg")
plt.ylabel("Height in cm")
plt.title("Average weight with respect to average height")
plt.legend(loc="lower right")

# Display the plot
plt.show()
```
