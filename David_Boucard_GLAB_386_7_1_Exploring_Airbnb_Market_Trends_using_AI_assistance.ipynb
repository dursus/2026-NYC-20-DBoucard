````python
# Import necessary packages such as pandas and numpy
import pandas as pd
import numpy as np
````

````python
# Import CSV for prices
airbnb_price = pd.read_csv('airbnb_price.csv')

# Import Excel file for room types
airbnb_room_type = pd.read_excel('airbnb_room_type.xlsx')

# Import TSV for review dates
airbnb_last_review = pd.read_csv('airbnb_last_review.tsv', sep='\t')
````

````python
# Merge airbnb_price and airbnb_room_type
# This merge combines pricing and location information with room type and description.
airbnb_merged = pd.merge(airbnb_price, airbnb_room_type, on='listing_id', how='inner')

# Merge the result with airbnb_last_review
# This step adds the host name and last review date to the combined DataFrame.
airbnb_merged = pd.merge(airbnb_merged, airbnb_last_review, on='listing_id', how='inner')

# Display the first 5 rows of the merged DataFrame
print("Merged DataFrame (first 5 rows):")
print(airbnb_merged.head())
````

````python
# Number of rows in each original dataset
rows_airbnb_price = airbnb_price.shape[0]
rows_airbnb_room_type = airbnb_room_type.shape[0]
rows_airbnb_last_review = airbnb_last_review.shape[0]

print(f"Rows in airbnb_price: {rows_airbnb_price}")
print(f"Rows in airbnb_room_type: {rows_airbnb_room_type}")
print(f"Rows in airbnb_last_review: {rows_airbnb_last_review}\n")

# First merge: airbnb_price and airbnb_room_type
initial_merge = pd.merge(airbnb_price, airbnb_room_type, on='listing_id', how='inner')
rows_initial_merge = initial_merge.shape[0]
records_lost_first_merge = (rows_airbnb_price + rows_airbnb_room_type) - rows_initial_merge

print(f"Rows after first merge (airbnb_price + airbnb_room_type): {rows_initial_merge}")
print(f"Records lost in first merge: {rows_airbnb_price - rows_initial_merge} (assuming airbnb_price was the base)\n")

# Second merge: with airbnb_last_review
rows_final_merge = airbnb_merged.shape[0] # airbnb_merged already contains the final merged data
records_lost_second_merge = rows_initial_merge - rows_final_merge

print(f"Rows after second merge (with airbnb_last_review): {rows_final_merge}")
print(f"Records lost in second merge: {records_lost_second_merge}\n")

# Total records lost
total_records_lost = (rows_airbnb_price + rows_airbnb_room_type + rows_airbnb_last_review) - (rows_initial_merge + (rows_airbnb_last_review - records_lost_second_merge))
# A more straightforward way for total_records_lost when using inner joins on common IDs:
# Since all original dataframes have the same number of unique listing_ids and they are the same set,
# and we used inner joins, no records should be lost if all listing_ids are present in all files.
# If there were differing listing_ids, the loss would be:
# (original_count_1 - final_merged_count) + (original_count_2 - final_merged_count) + ...
# Given the earlier check that 'Are listing_ids ... the same set? True', total_records_lost should be 0.
# Let's verify that with the unique counts we already printed:
# Unique listing_id in airbnb_price: 25209
# Unique listing_id in airbnb_room_type: 25209
# Unique listing_id in airbnb_last_review: 25209
# All inner joins on a common, identical set of IDs should result in 0 records lost relative to any single input.

print(f"Total unique listing_ids across all datasets: {rows_airbnb_price}")
print(f"Total records in final merged DataFrame: {rows_final_merge}")
print(f"Total records lost (relative to a single original dataset size if all IDs were in the final merge): {rows_airbnb_price - rows_final_merge}")
````

````python
# Clean the 'price' column by removing ' dollars' and converting to numeric
# Convert to string first to ensure .str accessor works, then remove ' dollars' and convert to float
airbnb_merged['price'] = airbnb_merged['price'].astype(str).str.replace(' dollars', '').astype(float)

print("\nDescriptive statistics for listing price:")
print(airbnb_merged['price'].describe())

# Extracting specific metrics for clarity
median_price = airbnb_merged['price'].median()
q1_price = airbnb_merged['price'].quantile(0.25)
q3_price = airbnb_merged['price'].quantile(0.75)
iqr_price = q3_price - q1_price

print(f"\nMedian Price: ${median_price:.2f}")
print(f"Q1 (25th percentile) Price: ${q1_price:.2f}")
print(f"Q3 (75th percentile) Price: ${q3_price:.2f}")
print(f"Interquartile Range (IQR) of Price: ${iqr_price:.2f}")
````

````python
# Calculate the share of listings for each room type
room_type_distribution = airbnb_merged['room_type'].value_counts(normalize=True) * 100
print("\nShare of listings by Room Type (%):\n")
print(room_type_distribution.round(2))

# Calculate the average price for each room type
average_price_by_room_type = airbnb_merged.groupby('room_type')['price'].mean().sort_values(ascending=False)
print("\nAverage Price by Room Type:\n")
print(average_price_by_room_type.apply(lambda x: f'${x:.2f}'))
````
