**Task 1: Data Type Objects in NumPy**
1. Create a NumPy array using np.array() function and specify the data type as 'int32.' Print the array and its data type.
2. Create another NumPy array using np.array() with the data type as 'float64.' Print the array and its data type.




```python
import numpy as np

# Create a NumPy array with 32-bit integers
int_array = np.array([1, 2, 3], dtype='int32')
print(int_array)
print(int_array.dtype)

# Create a NumPy array with 64-bit floats
float_array = np.array([1.0, 2.0, 3.0], dtype='float64')
print(float_array)
print(float_array.dtype)
```

**Task 2: One-Dimensional**
1. Create a one-dimensional NumPy array containing the integers from 1 to 5. Print the array.


```python
import numpy as np

# Create a NumPy array with integers 1 to 5
one_dim_array = np.array([1, 2, 3, 4, 5])
print(one_dim_array)
```

**Task 3: NumPy Operations**
1. Create two NumPy arrays, arr1 and arr2, with any value of your choice. Perform element-wise addition, subtraction, multiplication, and division between these arrays and print the results.
2. Calculate the dot product between arr1 and arr2. Print the result.
3. Use NumPy to calculate the mean, median, and standard deviation of arr1.


```python
import numpy as np

# Create two sample NumPy arrays to perform calculations
arr1 = np.array([12, 25, 42, 16, 32])
arr2 = np.array([4, 5, 6, 8, 8])

# Add, substract, multiply, and divide the two arrays
print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)

# Find the dot product
# (arr1[0]*arr2[0]) + (arr1[1]*arr2[1]) + (arr1[2]*arr2[2])...
print(np.dot(arr1, arr2))

# Find the mean, median, and standard deviation of the first array
print(np.mean(arr1))
print(np.median(arr1))
print(np.std(arr1))
```
